# making a basic figure with matplotlib

First, import relevant libraries and packages

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy, PIL, requests
```

Our first plot is a scatter plot of 50 random points (x ranges from 0 to 49 and y is a random number between 1 and 100). Remember to set a random seed for reproducibilty of your code.

```python
np.random.seed(613)

x= np.arange(50)
y= np.random.randint(0,100,50)

fig,ax = plt.subplots(figsize=(5,3)) #figsize specifies the size of the figure
ax.scatter(x,y)
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.barh(x,y) #this makes a bar graph
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y) #this produces a line graph
```

```python
fig,ax = plt.subplots(figsize=(5,3))

ax.hist(y) #this produces a histogram
```

```python
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total growth over time') #adding a title and axis label to the figures
ax.set_ylabel('total growth')
ax.set_xlabel('years since start')

fig.tight_layout()
```

```python
#remember that font is a dictionary
font1 = {'family': 'sans-serif','color':'blue','size':20}

font2 = {'family': 'monospace','color':'green','size':20}
```

```python
fig,ax = plt.subplots(figsize= (5,3))
ax.plot(x,y)

ax.set_title('total growth over time',fontdict = font1, loc ='left')
ax.set_ylabel('total growth',fontdict = font2)
ax.set_xlabel('years since start',fontdict = font2)
```

# modifying our data points


```python
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x,y,
           marker = '*',
           color = 'indigo')
```

```python
fig,ax = plt.subplots(figsize=(5,3))

ax.plot(x,y,
        marker = '*',
        #color = 'indigo',
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2)
```

```python
fig,ax = plt.subplots(figsize=(5,3))

ax.plot(x,y,
        marker = '*',
        markersize = 15,
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')

ax.grid(axis ='y', color = 'blue', linewidth = 2, linestyle = '-.')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Reds", fill=True)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length,
            cmap="Blues", fill=True, bw_adjust=0.5)
plt.show()
```
